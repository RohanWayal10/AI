# PR6: Regression Techniques (CO2, CO3)

**Aim:** Implement Linear Regression and Polynomial Regression using the California Housing dataset.

**Evaluation Metrics:**
- Mean Squared Error (MSE)
- Mean Absolute Error (MAE)
- R² Score

Finally, compare the performance of both regression techniques.


## Step 1: Import Libraries


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import fetch_california_housing
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
```

## Step 2: Load California Housing Dataset


```python
housing = fetch_california_housing()

df = pd.DataFrame(housing.data, columns=housing.feature_names)
df['Price'] = housing.target

df.head()
```

## Step 3: Explore the Dataset


```python
print("Dataset Shape:", df.shape)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nStatistical Summary:")
display(df.describe())
```

## Step 4: Select Feature and Target

For simple visualization, **MedInc (Median Income)** is used as the input feature and **Price** as the target variable.


```python
X = df[['MedInc']]
y = df['Price']

print("Input Feature:")
display(X.head())

print("Target:")
display(y.head())
```

## Step 5: Split Dataset into Training and Testing Sets


```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42
)

print("Training Samples:", len(X_train))
print("Testing Samples:", len(X_test))
```

## Step 6: Implement Linear Regression


```python
linear_model = LinearRegression()
linear_model.fit(X_train, y_train)

y_pred_linear = linear_model.predict(X_test)
```

## Step 7: Evaluate Linear Regression


```python
linear_mse = mean_squared_error(y_test, y_pred_linear)
linear_mae = mean_absolute_error(y_test, y_pred_linear)
linear_r2 = r2_score(y_test, y_pred_linear)

print("Linear Regression Results")
print("-------------------------")
print("MSE      :", linear_mse)
print("MAE      :", linear_mae)
print("R² Score :", linear_r2)
```

## Step 8: Implement Polynomial Regression (Degree = 2)


```python
poly = PolynomialFeatures(degree=2)

X_train_poly = poly.fit_transform(X_train)
X_test_poly = poly.transform(X_test)

poly_model = LinearRegression()
poly_model.fit(X_train_poly, y_train)

y_pred_poly = poly_model.predict(X_test_poly)
```

## Step 9: Evaluate Polynomial Regression


```python
poly_mse = mean_squared_error(y_test, y_pred_poly)
poly_mae = mean_absolute_error(y_test, y_pred_poly)
poly_r2 = r2_score(y_test, y_pred_poly)

print("Polynomial Regression Results")
print("-----------------------------")
print("MSE      :", poly_mse)
print("MAE      :", poly_mae)
print("R² Score :", poly_r2)
```

## Step 10: Compare Both Models


```python
results = pd.DataFrame({
    'Model': ['Linear Regression', 'Polynomial Regression'],
    'MSE': [linear_mse, poly_mse],
    'MAE': [linear_mae, poly_mae],
    'R² Score': [linear_r2, poly_r2]
})

results
```

## Step 11: Visualize Linear Regression


```python
plt.figure(figsize=(8, 5))
plt.scatter(X_test, y_test, alpha=0.4, label='Actual Data')
plt.scatter(X_test, y_pred_linear, alpha=0.4, label='Predicted Data')

plt.xlabel('Median Income')
plt.ylabel('House Price')
plt.title('Linear Regression')
plt.legend()
plt.show()
```

## Step 12: Visualize Polynomial Regression


```python
X_plot = np.linspace(
    X['MedInc'].min(),
    X['MedInc'].max(),
    500
).reshape(-1, 1)

X_plot_df = pd.DataFrame(X_plot, columns=['MedInc'])
X_plot_poly = poly.transform(X_plot_df)
y_plot_poly = poly_model.predict(X_plot_poly)

plt.figure(figsize=(8, 5))
plt.scatter(X_test, y_test, alpha=0.3, label='Actual Data')
plt.plot(X_plot, y_plot_poly, linewidth=2, label='Polynomial Regression')

plt.xlabel('Median Income')
plt.ylabel('House Price')
plt.title('Polynomial Regression (Degree = 2)')
plt.legend()
plt.show()
```

## Step 13: Final Comparison


```python
print("Final Model Comparison")
print("=" * 50)

print("\nLinear Regression")
print("MSE :", round(linear_mse, 4))
print("MAE :", round(linear_mae, 4))
print("R²  :", round(linear_r2, 4))

print("\nPolynomial Regression")
print("MSE :", round(poly_mse, 4))
print("MAE :", round(poly_mae, 4))
print("R²  :", round(poly_r2, 4))

if poly_r2 > linear_r2:
    print("\nPolynomial Regression performs better.")
else:
    print("\nLinear Regression performs better.")
```

## Conclusion

Linear Regression and Polynomial Regression were successfully implemented using the California Housing dataset.

Both models were evaluated using **MSE, MAE, and R² Score**. A model with lower MSE and MAE and a higher R² Score is considered to provide better predictive performance.
