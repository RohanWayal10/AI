## 2. Theory

An **intelligent agent** perceives its environment through sensors and takes actions through actuators to achieve a goal.

A **rule-based agent** selects actions using predefined condition-action rules. For example, if a room is dark and occupied, the agent turns on the light.

### PEAS Representation

- **P — Performance Measure:** Criteria used to evaluate the agent's success.
- **E — Environment:** The world in which the agent operates.
- **A — Actuators:** Components through which the agent performs actions.
- **S — Sensors:** Components that provide information about the environment.

## 3. PEAS Representation: Smart Home Controller

```python
peas = {
    "Performance Measure": [
        "Maintain comfortable temperature",
        "Reduce unnecessary energy consumption",
        "Provide lighting when the room is occupied and dark"
    ],
    "Environment": [
        "Smart home",
        "Living room",
        "Indoor temperature and light conditions",
        "Room occupancy"
    ],
    "Actuators": [
        "Heater or air conditioner",
        "Light switch"
    ],
    "Sensors": [
        "Temperature sensor",
        "Light sensor",
        "Occupancy sensor"
    ]
}

for component, details in peas.items():
    print(f"\n{component}:")
    for detail in details:
        print(" -", detail)
```

## 4. Define the Intelligent Agent

```python
def smart_home_agent(temperature, light_level, occupied):
    """Return actions based on temperature, brightness, and occupancy.

    temperature: indoor temperature in degrees Celsius
    light_level: 'dark' or 'bright'
    occupied: True if someone is in the room, otherwise False
    """
    actions = []

    # Temperature control
    if temperature < 18:
        actions.append("Turn HEATER ON")
    elif temperature > 28:
        actions.append("Turn AIR CONDITIONER ON")
    else:
        actions.append("Keep temperature control OFF")

    # Lighting control
    if occupied and light_level == "dark":
        actions.append("Turn LIGHT ON")
    else:
        actions.append("Turn LIGHT OFF")

    return actions

print("Smart Home Agent function created successfully.")
```

## 5. Test the Agent with Different Scenarios

```python
scenarios = [
    {
        "Scenario": "Cold room, occupied, dark",
        "temperature": 16,
        "light_level": "dark",
        "occupied": True
    },
    {
        "Scenario": "Hot room, occupied, bright",
        "temperature": 31,
        "light_level": "bright",
        "occupied": True
    },
    {
        "Scenario": "Comfortable room, unoccupied, dark",
        "temperature": 23,
        "light_level": "dark",
        "occupied": False
    },
    {
        "Scenario": "Comfortable room, occupied, dark",
        "temperature": 23,
        "light_level": "dark",
        "occupied": True
    }
]

for i, scenario in enumerate(scenarios, start=1):
    actions = smart_home_agent(
        scenario["temperature"],
        scenario["light_level"],
        scenario["occupied"]
    )

    print(f"\nScenario {i}: {scenario['Scenario']}")
    print(f"Temperature: {scenario['temperature']} °C")
    print(f"Light level: {scenario['light_level']}")
    print(f"Room occupied: {scenario['occupied']}")
    print("Agent actions:")
    for action in actions:
        print(" -", action)
```

## 6. Interactive Agent Demonstration

```python
# Change these values to test your own situation
current_temperature = 25
current_light_level = "dark"  # Choose 'dark' or 'bright'
room_occupied = True          # Choose True or False

print("Sensor readings:")
print("Temperature:", current_temperature, "°C")
print("Light level:", current_light_level)
print("Room occupied:", room_occupied)

print("\nAgent response:")
for action in smart_home_agent(
    current_temperature,
    current_light_level,
    room_occupied
):
    print(" -", action)
```

## 7. Observe Agent Behavior

```python
import pandas as pd

observations = []

for scenario in scenarios:
    actions = smart_home_agent(
        scenario["temperature"],
        scenario["light_level"],
        scenario["occupied"]
    )
    observations.append({
        "Scenario": scenario["Scenario"],
        "Temperature (°C)": scenario["temperature"],
        "Light Level": scenario["light_level"],
        "Occupied": scenario["occupied"],
        "Agent Actions": "; ".join(actions)
    })

pd.DataFrame(observations)
```
